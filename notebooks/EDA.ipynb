```python
import glob
import pandas as pd
import os

# Removed the '..' since VS Code is running this from your workspace root!
all_files = glob.glob(os.path.join('data', 'raw', 'bidmc', '*_Numerics.csv'))

dfs = []
for f in all_files:
    # skipinitialspace=True removes the leading spaces in " HR" and " SpO2"
    temp = pd.read_csv(f, skipinitialspace=True)[['Time [s]', 'HR', 'SpO2']]
    
    # Rename 'Time [s]' to just 'Time'
    temp.rename(columns={'Time [s]': 'Time'}, inplace=True)
    
    # Extract the patient ID safely
    temp['patient_id'] = os.path.basename(f).split('bidmc')[-1].replace('_Numerics.csv', '').strip('_')
    dfs.append(temp)

if dfs:
    df_all = pd.concat(dfs, ignore_index=True)
    print(f"Total BIDMC records: {len(df_all)}")
    display(df_all.head())
else:
    print("No CSV files found. Make sure the path is correct!")
```
